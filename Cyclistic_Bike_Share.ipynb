```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Load the Excel file
cyclistic_df = pd.read_csv(r'D:\Merit America\Case Study 1\cyclistic_Bike.csv')
cyclistic_df.head()
```

```python
cyclistic_df.info()
cyclistic_df.describe()
```

```python
cyclistic_df.columns
```

```python
cyclistic_df['started_at'] = pd.to_datetime(cyclistic_df['started_at'])
cyclistic_df['ended_at'] = pd.to_datetime(cyclistic_df['ended_at'])
```

```python
cyclistic_df.head()
```

```python
cyclistic_df['ride_length'] = (cyclistic_df['ended_at'] - cyclistic_df['started_at']).dt.total_seconds() / 60
cyclistic_df.head()
```

```python
cyclistic_df = cyclistic_df[cyclistic_df['ride_length'] >= 0]
cyclistic_df.head()
```

```python
cyclistic_df['day_of_week'] = cyclistic_df['started_at'].dt.day_name()
cyclistic_df['hour'] = cyclistic_df['started_at'].dt.hour
cyclistic_df['month'] = cyclistic_df['started_at'].dt.month
cyclistic_df['month_name'] = cyclistic_df['started_at'].dt.month_name()
cyclistic_df['is_weekend'] = cyclistic_df['day_of_week'].isin(['Saturday', 'Sunday'])
cyclistic_df.head()
```
